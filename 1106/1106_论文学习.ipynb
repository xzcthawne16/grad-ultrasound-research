## 组会汇报要点：Perdios 2021（TUFFC）

> 论文：CNN-Based Image Reconstruction Method for Ultrafast Ultrasound Imaging
> 作者：Perdios et al., EPFL · 期刊：IEEE TUFFC · DOI: 10.1109/TUFFC.2021.3131383

### 1. 它解决什么
单次平面波（低质 LQ）经 CNN 逐像素重建，逼近 192 次合成孔径（金标准 SA / 高质 HQ）的质量，
从而在保持超快成像帧率的同时，去除衍射伪影（光栅瓣、旁瓣、边缘波）。

### 2. 关键结果（含诚实边界）
- 对比度大幅改善、全局 PSNR/SSIM 甚至超过金标准 SA，动态范围 > 60 dB；
- 但「亮点反射体」横向分辨率显著改善（446.6 → 322.5 µm），
  而「散斑模式」横向分辨率几乎无改善（262.1 → 251.1 µm）——这是作者自己坦承的短板；
- 深层改善不明显，颈动脉内膜等精细结构未能准确恢复。

### 3. 对我项目的启示
- 选「可比、有物理含义」的指标：对比度、亮点 FWHM、散斑 ACF，不无脑套 PSNR/SSIM；
- 讲结果时严格区分「亮点 vs 散斑」两类目标；
- 恢复映射绑定特定成像配置，换探头/参数需重训，泛化受限。
