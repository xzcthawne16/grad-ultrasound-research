-   1.开头三问
    -   1.1昨天 K 曲线得出了什么？
        -   参照系错误，点恢复率用了「峰值归一化」（img/max），这会把绝对幅值的丢失掩盖掉；gCNR 又有「分布可分度」的陷阱。所以昨天判定：点恢复率、gCNR 这两个指标参照系有误，不采信，只有 PSNR/SSIM 可信。
        -   仿真太理想：因为无噪声无衰减，每多加一个角度都有真实增益，导致 PSNR 一直涨到 K≈49 才饱和——远大于真实超声的 10~20。
    -   1.2今天在整条链的哪一环？
        -   整条验证链：数据→波束形成→K档复合→指标→临界K值
        -   数据环节逼近真实：给仿真加上电子噪声 + 深度衰减，让链不再「太理想」，看临界 K 是否回落到 10~20（主任务 1A）。
        -   指标环节补上「保组织」这一维：昨天判了「点恢复率/gCNR 参照系错误」，今天用正确参照系（理想高斯点图，不是 K=75 图）​重新算「点恢复率」，回答「少角度到底有没有把真实点抹掉」（主任务 1B）。
    -   1.3今天做完，我能多做哪一件具体的事？
        -   「真实超声里，用 1/N 的发射（N≈3~7，即 K≈10~20）就能逼近满配质量」——帧率故事从「自仿真的 49」变成「真实量级的 10~20」，结论变硬。
            -   「1/N 发射」= 用「满配发射次数」的 1/N，就达到接近满配的画质。​ N 越大，说明你省得越多、帧率提得越狠。
        -   「少角度复合不丢真实点目标」——这是「去伪影 vs 保组织」权衡的跨角度证据，直接喂给你的 research gap。
