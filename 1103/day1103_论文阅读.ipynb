    -   Visual assessment of the test phantom images (Fig. 4) and the metrics obtained (Table II) confirm that the proposed image reconstruction method significantly improves the image quality compared with LQ using any of the trained CNNs.

        -   对具有相同容量的CNN（即MSE-16、MAE-16和MSLA-E-16）进行全球比较，表明所提出的HDR敏感型MSLA-E损失具有优越性。

        -   MSE-16 和 MAE-16 在低回声包含体（图4(f)和4(g)中的ΩI）中均存在明显的“暗区伪影”[55]，而MSLAE-16 和 MSLAE-32 对该包含体的修复更为准确。

-   标题：Interpretable Deep Transfer Learning for Breast Ultrasound Cancer Detection: A Multi-Dataset Study
    -   可解释的深度迁移学习在乳腺超声癌症检测中的应用：多数据集研究
-   Abstract：
    -   Breast cancer remains a leading cause of cancerrelated mortality among women worldwide. Ultrasound imaging, widely used due to its safety and cost-effectiveness, plays a key role in early detection, especially in patients with dense breast tissue.
        -   乳腺癌仍是全球女性癌症相关死亡的主要原因之一。由于安全性和经济性，超声成像被广泛使用，在早期检测中发挥着关键作用，尤其适用于乳腺组织较密的患者。
    -   This paper presents a comprehensive study on the application of machine learning and deep learning techniques for breast cancer classification using ultrasound images.
        -   本文对利用超声图像进行乳腺癌分类的机器学习和深度学习技术应用进行了全面研究。
    -   Using datasets such as BUSI, BUS-BRA, and BrEaST-Lesions USG, we evaluate classical machine learning models (SVM, KNN) and deep convolutional neural networks (ResNet-18, EfficientNetB0, GoogLeNet).
        -   我们使用BUSI、BUS-BRA和BrEaST-Lesions USG等数据集，评估了经典的机器学习模型（SVM、KNN）以及深度卷积神经网络（ResNet-18、EfficientNetB0、GoogLeNet）。
    -   Experimental results show that ResNet-18 achieves the highest accuracy (99.7%) and perfect sensitivity for malignant lesions.
        -   实验结果表明，ResNet-18对恶性病变达到了最高的准确率（99.7%）和完美的灵敏度。
    -   Classical ML models, though outperformed by CNNs, achieve competitive performance when enhanced with deep feature extraction.
        -   经典机器学习模型虽然在性能上被卷积神经网络（CNN）超越，但通过增强深度特征提取后仍能实现具有竞争力的性能。
    -   Grad-CAM visualizations further improve model transparency by highlighting diagnostically relevant image regions.
        -   Grad-CAM可视化通过突出显示具有诊断意义的图像区域，进一步提升了模型的透明度。
    -   These findings support the integration of AI-based diagnostic tools into clinical workflows and demonstrate the feasibility of deploying high-performing, interpretable systems for ultrasound-based breast cancer detection.
        -   这些发现支持将基于人工智能的诊断工具整合到临床工作流程中，并证明了在超声乳腺癌检测中部署高性能、可解释系统的可行性。
-   Conclusion：
    -   In this study, ultrasound-based breast cancer detection was evaluated using machine learning and deep learning on publicly available datasets, notably BUSI and BUS-BRA.
        -   本研究利用机器学习和深度学习方法，对公开可用的数据集（特别是BUSI和BUS-BRA）进行了基于超声的乳腺癌检测评估。
    -   Classical classifiers (SVM, KNN) were systematically compared with transfer-learned CNNs (ResNet-18, EfficientNet-B0).
        -   经典分类器（SVM、KNN）与通过迁移学习训练的CNN（ResNet-18、EfficientNet-B0）进行了系统性比较。
    -   Deep models clearly outperformed traditional approaches; ResNet-18 achieved 99.7% accuracy and perfect malignant recall.
        -   深度模型明显优于传统方法; ResNet-18实现了99.7%的准确率和完美的恶性召回。
    -   The results confirm the feasibility of AI tools to assist radiologists with high sensitivity and precision. Explainability via Grad-CAM highlighted diagnostically relevant regions, strengthening clinical trust.
        -   结果证实了人工智能工具在高灵敏度和高精度方面辅助放射科医生的可行性。通过Grad-CAM实现的可解释性突出了具有诊断相关性的区域，从而增强了临床信任。
    -   Nonetheless, real-world deployment requires multi-center validation, integrated lesion localization, and careful assessment of generalizability across scanners and patient demographics.
        -   然而，实际应用需要多中心验证、整合性病灶定位，以及对不同扫描仪和患者人群的泛化性进行仔细评估。
    -   Future work targets end-toend diagnostic pipelines, multimodal fusion, and prospective evaluation.
        -   未来的工作重点包括端到端的诊断流程、多模态融合以及前瞻性评估。
    -   Overall, the study underscores clinically oriented, interpretable AI and positions ultrasound as a practical, costeffective platform for improving early detection, especially in resource-constrained settings worldwide.
        -   总体而言，该研究强调了以临床为导向、易于解读的人工智能，并将超声检查定位为一种实用且经济高效的平台，有助于提高早期检测能力，尤其在世界各地资源有限的地区。
