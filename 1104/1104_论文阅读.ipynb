-   标题：Learning a Clinically-Relevant Concept Bottleneck for Lesion Detection in Breast Ultrasound
    -   学习乳腺超声中病变检测的临床相关概念瓶颈
- Abstract：
  - Detecting and classifying lesions in breast ultrasound images is a promising application of artificial intelligence (AI) for reducing the burden of cancer in regions with limited access to mammography.
    - 在乳腺超声图像中检测和分类病灶，是人工智能（AI）在乳腺癌筛查方面的一项有前景的应用，有助于减轻那些难以获得乳腺X线检查地区的癌症负担。
  - Such AI systems are more likely to be useful in a clinical setting if their predictions can be explained to a radiologist.
    - 如果这些人工智能系统能够向放射科医生解释其预测结果，那么它们在临床环境中就更有可能发挥作用。
  - This work proposes an explainable AI model that provides interpretable predictions using a standard lexicon from the American College of Radiology’s Breast Imaging and Reporting Data System (BI-RADS).
    - 本研究提出了一种可解释的AI模型，利用美国放射学学院乳腺影像与报告数据系统（BI-RADS）中的标准术语表，提供可解释的预测结果。
  - The model is a deep neural network featuring a concept bottleneck layer in which known BIRADS features are predicted before making a final cancer classification.
    - 该模型是一种深度神经网络，包含一个瓶颈层概念，在做出最终的癌症分类之前，先预测已知的BIRADS特征。
  - This enables radiologists to easily review the predictions of the AI system and potentially fix errors in real time by modifying the concept predictions.
    - 这使放射科医生能够轻松地审查AI系统的预测结果，并通过修改概念预测来实时纠正可能存在的错误。
  - In experiments, a model is developed on 8,854 images from 994 women with expert annotations and histological cancer labels.
    - 在实验中，基于994名女性的8,854张图像，结合专家标注和组织学癌症标签，构建了一个模型。
  - The model outperforms state-of-the-art lesion detection frameworks with 48.9 average precision on the held-out testing set, and for cancer classification, concept intervention is shown to increase performance from 0.876 to 0.885 area under the receiver operating characteristic curve.
    - 该模型在预留测试集上平均精确度达到48.9，优于当前最先进的病灶检测框架；在癌症分类方面，概念干预使受试者操作特征曲线下的面积从0.876提升至0.885。
  - Training and evaluation code is available at https://github.com/hawaii-ai/bus-cbm.
    - 训练和评估代码可从 https://github.com/hawaii-ai/bus-cbm 获取。
- Conclusion：
  - To enhance interpretability of cancer status classification of lesions in BUS, while simultaneously providing automatic lesion delineation, we propose BI-RADS CBM.
    - 为了提高BUS中病灶癌变状态分类的可解释性，同时实现病灶的自动勾画，我们提出BI-RADS CBM。
  - Experiments on a large internal dataset of BUS imaging with expert annotations and histological cancer labels demonstrate that an AI model can be both accurate and interpretable.
    - 在包含专家标注和组织学癌症标签的大型内部BUS成像数据集上进行的实验表明，人工智能模型既准确又可解释。
  - We further demonstrate for the first time that BI-RADS concept intervention is possible and increases cancer classification performance.
    - 我们首次进一步证明，BI-RADS概念干预是可行的，并能提升癌症分类的性能。
  - BI-RADS CBM presents an XAI solution for lesion detection, description, and classification from BUS suitable for use in limited-resource scenarios to stretch limited radiological resources.
    - BI-RADS CBM 提供了一种可解释的 AI 解决方案，用于从超声图像中检测、描述和分类病灶，适用于资源有限的场景，以充分利用有限的放射学资源。
- 启发：乳腺超声任务强调「临床相关、按病例组织」——提醒我们 BUS-BRA 训练/验证必须按病例（patient-level）划分，否则同一病灶多张图会泄漏到验证集
