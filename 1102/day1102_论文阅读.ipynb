  - For each training experiment, kernel weights were initialized using the well-known Glorot initialization [49] with a uniform distribution, and biases were initialized to zero.

  - Model parameters were optimized using the Adam optimizer [50] with a learning rate of 5 × 10−5.

  - Mini-batch learning was deployed with a batch size of two. The training set consisted of 30 000 image pairs; a size motivated by a dedicated study to prevent training experiments from overfitting (Supplementary Material, Section S-III-F).
    - 部署了小批量学习，批次大小为两个。训练集由3万对图像组成；这个大小是由一项专门的研究推动的，目的是防止训练实验过拟合(补充材料，S-III-F部分)。

    - 在每个时期之后执行训练集的完全随机洗牌。
  - Neither training regularization (e.g., dropout or weight regularization) nor data augmentation was used.
    - 既没有使用训练正则化(例如，辍学或权重正则化)，也没有使用数据增强。
  - To fulfill the downsampling restrictions imposed by the proposed CNN architecture (Fig. 1), input images were zero padded symmetrically to the closest supported image shape (i.e., 608 × 1600), and cropped to their original size after inference.

  - To monitor and evaluate the performance of each training experiment, we used a validation set of 500 image pairs.
    - 为了监控和评估每个训练实验的性能，我们使用了500个图像对的验证集。
  - Both the peak signal-to-noise ratio (PSNR) and the structural similarity (SSIM) index [51] were computed at each validation step (i.e., every 1000 iterations).
    - 在每个验证步骤(即，每1000次迭代)计算峰值信噪比(PSNR)和结构相似性(SSIM)指数[51]。
  - These metrics were evaluated on B-mode representations between −62 and +36 dB (confidence interval discussed in Section III-B3), and were averaged over the entire validation set.
    - 这些指标在−62和+36分贝(在第III-B3节中讨论的可信区间)之间的B模式表示法上进行评估，并在整个验证集上求平均值。
  - The B-mode SSIM correlated well with visual assessments for evaluating the overall quality of recovered images, and was used to select the best performing CNN instance among the 500 validation steps of each training experiment.
    - B模式SSIM与用于评估恢复图像的整体质量的视觉评估具有很好的相关性，并被用于从每个训练实验的500个验证步骤中选择执行最好的CNN实例。
  - For comparison purposes, a fixed random seed was used for initializing kernel weights, identical training set shufflings were performed, and the same validation set was used for each training experiment.

- Abstract：
  - Prenatal ultrasound evaluates fetal growth and detects congenital abnormalities during pregnancy, but the examination of ultrasound images by radiologists requires expertise and sophisticated equipment, which would otherwise fail to improve the rate of identifying specific types of fetal central nervous system (CNS) abnormalities and result in unnecessary patient examinations.
    - 产前超声可以评估胎儿的发育，并在怀孕期间检测先天异常，但放射科医生对超声图像的检查需要专业知识和复杂的设备，否则将无法提高对特定类型胎儿中枢神经系统(CNS)异常的识别率，并导致不必要的患者检查。
  - We construct a deep learning model to improve the overall accuracy of the diagnosis of fetal cranial anomalies to aid prenatal diagnosis.
    - 我们构建了一个深度学习模型，以提高胎儿颅骨异常诊断的整体准确性，以辅助产前诊断。
  - In our collected multi-center dataset of fetal craniocerebral anomalies covering four typical anomalies of the fetal central nervous system (CNS): anencephaly, encephalocele (including meningocele), holoprosencephaly, and rachischisis, patient-level prediction accuracy reaches 94.5%, with an AUROC value of 99.3%.
    - 在我们收集的涵盖无脑、脑膨出(包括脑膜膨出)、全前脑和脊柱裂四种典型胎儿中枢神经系统异常的多中心胎儿颅脑异常数据集中，患者水平的预测准确率达到94.5%，AUROC值为99.3%。
  - In the subgroup analyzes, our model is applicable to the entire gestational period, with good identification of fetal anomaly types for any gestational period.
    - 在子群分析中，我们的模型适用于整个妊娠期，对任何妊娠期的胎儿异常类型都有很好的识别能力。
  - Heatmaps superimposed on the ultrasound images not only provide a visual interpretation for the algorithm but also provide an intuitive visual aid to the physician by highlighting key areas that need to be reviewed, helping the physician to quickly identify and validate key areas.
    - 叠加在超声图像上的热图不仅为算法提供了直观的解释，还通过突出显示需要审查的关键区域为医生提供了直观的视觉帮助，帮助医生快速识别和验证关键区域。
